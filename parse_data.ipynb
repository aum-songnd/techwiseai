```python
import json
```

```python
with open('laptops_cleaned_for_import.json', "r", encoding="utf-8") as f:
    data = json.load(f)
names = [products['name'] for products in data['products']]
```

```python
with open('laptop_names.txt', 'w', encoding='utf-8') as f:
    for name in names:
        f.write(name + '\n')
```

```python
with open('laptop.json', 'r', encoding='utf-8') as f:
    laptops = json.load(f)
```

```python
filter = []
for laptop in laptops:
    if laptop['name'] in names:
        filter.append(laptop)
len(filter)
```

```python
with open('laptop_clean.json', "w", encoding="utf-8") as file:
    json.dump(filter, file, ensure_ascii=False, indent=2)
```

```python
print("Số lượng laptop đã lọc:", len(filter))
print("Số lượng laptop ban đầu:", len(laptops))
```
